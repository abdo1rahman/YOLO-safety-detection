```python
from collections import Counter
import math

def validate_yolo_labels(label_dir, num_classes):

    stats = {
        "total_files": 0,
        "empty_files": 0,
        "total_objects": 0,
        "class_counts": Counter(),
        "errors": []
    }

    label_files = list(Path(label_dir).glob("*.txt"))

    stats["total_files"] = len(label_files)

    for label_file in label_files:

        try:
            lines = label_file.read_text().strip().splitlines()

            # Empty annotation files are allowed.
            if not lines or lines == [""]:
                stats["empty_files"] += 1
                continue

            for line_number, line in enumerate(lines, start=1):

                values = line.split()

                if len(values) != 5:
                    raise ValueError(
                        f"Expected 5 values, found {len(values)}"
                    )

                class_value = float(values[0])

                if not math.isfinite(class_value):
                    raise ValueError("Non-finite class ID")

                if not class_value.is_integer():
                    raise ValueError("Class ID must be an integer")

                class_id = int(class_value)

                if not 0 <= class_id < num_classes:
                    raise ValueError(
                        f"Invalid class ID: {class_id}"
                    )

                x_center, y_center, width, height = map(
                    float, values[1:]
                )

                coordinates = [
                    x_center, y_center, width, height
                ]

                if not all(math.isfinite(v) for v in coordinates):
                    raise ValueError("Non-finite coordinates")

                if not all(0 <= v <= 1 for v in coordinates):
                    raise ValueError(
                        "Coordinates outside normalized range"
                    )

                if width <= 0 or height <= 0:
                    raise ValueError(
                        "Bounding-box width and height must be positive"
                    )

                # Verify that the bounding box stays within the image.
                x_min = x_center - width / 2
                x_max = x_center + width / 2
                y_min = y_center - height / 2
                y_max = y_center + height / 2

                tolerance = 1e-6

                if (
                    x_min < -tolerance
                    or y_min < -tolerance
                    or x_max > 1 + tolerance
                    or y_max > 1 + tolerance
                ):
                    raise ValueError(
                        "Bounding box extends outside image boundaries"
                    )

                stats["class_counts"][class_id] += 1
                stats["total_objects"] += 1

        except Exception as error:

            stats["errors"].append({
                "file": label_file.name,
                "error": str(error)
            })

    return stats
```

```python
split_stats = {}

for split in ["train", "valid", "test"]:

    print(f"\n{'=' * 50}")
    print(f"VALIDATING: {split.upper()}")
    print("=" * 50)

    label_dir = DATASET_ROOT / split / "labels"

    stats = validate_yolo_labels(
        label_dir=label_dir,
        num_classes=NUM_CLASSES
    )

    split_stats[split] = stats

    print("Annotation files:", stats["total_files"])
    print("Empty annotation files:", stats["empty_files"])
    print("Total objects:", stats["total_objects"])
    print("Invalid annotations:", len(stats["errors"]))

    print("\nClass distribution:")

    for class_id, class_name in enumerate(CLASS_NAMES):
        count = stats["class_counts"][class_id]

        print(f"{class_id:2d} | {class_name:15s} | {count:,}")

    if stats["errors"]:
        print("\nFirst 10 errors:")

        for error in stats["errors"][:10]:
            print(error)
```

```python
from PIL import Image
from collections import Counter

for split in ["train", "valid", "test"]:

    image_dir = DATASET_ROOT / split / "images"

    dimensions = Counter()
    formats = Counter()
    corrupted = []

    images = [
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in IMAGE_EXTENSIONS
    ]

    for image_path in images:

        try:
            with Image.open(image_path) as image:
                dimensions[image.size] += 1
                formats[image.format] += 1

        except Exception as error:
            corrupted.append((image_path.name, str(error)))

    print(f"\n{split.upper()}")
    print("Image dimensions:", dimensions.most_common(10))
    print("Formats:", formats)
    print("Corrupted images:", len(corrupted))
```

```python
import hashlib
from collections import defaultdict

def calculate_md5(file_path):

    md5 = hashlib.md5()

    with open(file_path, "rb") as file:
        for chunk in iter(lambda: file.read(8192), b""):
            md5.update(chunk)

    return md5.hexdigest()


hash_to_splits = defaultdict(set)
hash_to_files = defaultdict(list)

for split in ["train", "valid", "test"]:

    image_dir = DATASET_ROOT / split / "images"

    for image_path in image_dir.iterdir():

        if image_path.suffix.lower() not in IMAGE_EXTENSIONS:
            continue

        image_hash = calculate_md5(image_path)

        hash_to_splits[image_hash].add(split)
        hash_to_files[image_hash].append(image_path.name)


cross_split_duplicates = {
    image_hash: files
    for image_hash, files in hash_to_files.items()
    if len(hash_to_splits[image_hash]) > 1
}

print("Unique image hashes:", len(hash_to_files))
print("Exact duplicate groups across splits:",
      len(cross_split_duplicates))

for files in list(cross_split_duplicates.values())[:10]:
    print(files)
```
